```python
# Import libraries and load the Titanic dataset

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Set theme
sns.set_theme(style="darkgrid")

# Load Titanic datasets
titanic = sns.load_dataset("titanic")
```

```python
# 1. Countplot - Passenger Count by Embarked Port

plt.figure(figsize=(8,5))
sns.countplot(data=titanic, x='embark_town', hue='sex', palette='viridis')

plt.title('Count of passengers by embarked port')
plt.xlabel('Embarked Port')
plt.ylabel('Number of passengers')
plt.legend(title='Gender')

plt.show()
```

Observation:

Southampton accounts for roughly 70% of boardings, far more than Cherbourg or
Queenstown.
Males outnumber females at Southampton and Cherbourg.
Queenstown is the only port where the two are close to even.

```python
# 2. Swarmplot: Age Distribution across Classes

plt.figure(figsize=(9,5))
sns.swarmplot(data=titanic, x='pclass', y='age', hue='sex', palette='viridis')

plt.title('Age Distribution across Classes')
plt.xlabel('Passenger Class')
plt.ylabel('Age')
plt.legend(title='Gender')

plt.show()
```

Observation:

Passengers of different ages travelled in all three classes. Most passengers were adults, and children were present in all passenger classes.

```python
# 3. Pie Chart: Gender Distribution
passenger_gender_count = titanic['sex'].value_counts()
explode = (0.05, 0)

plt.figure(figsize=(8,8))
plt.pie(passenger_gender_count,
        labels=passenger_gender_count.index,
        colors=['lightcoral', 'lightskyblue'],
        autopct='%1.1f%%',
        explode=explode, startangle=90)
plt.title('Gender Distribution')
plt.axis('equal')

plt.show()
```

Obeservation:

About 65% of passengers were male and 35% female.
The exploded female slice highlights the smaller group.

```python
# 4. Heatmap: Correlation Matrix

plt.figure(figsize=(8,6))
corr_data = ['age', 'fare', 'pclass','sibsp', 'parch', 'survived']
corr_matrix = titanic[corr_data].corr()

sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Matrix')
plt.tight_layout()

plt.show()
```

Observation:

pclass and fare are strongly negatively correlated (about -0.55), since lower class numbers mean higher fares.
survived correlates negatively with pclass (about -0.34) and positively with fare (about 0.26), so higher-class, higher-paying passengers survived more often.
sibsp and parch are mildly positively correlated with each other, because larger families travel together.
Age correlates weakly negatively with pclass (older passengers were more often in first class).

```python
# 5. Violin Plot: Fare by Class and Gender

plt.figure(figsize=(8,5))
sns.violinplot(data = titanic, x='pclass', y='fare', hue='sex', split=True, palette='muted', cut=0)

plt.title("Fare Distribution by Class and Gender")
plt.xlabel("Passenger Class")
plt.ylabel("Fare")
plt.legend(title="Gender")
plt.tight_layout()

plt.show()
```

Observation:

First-class passengers generally paid higher fares than second- and third-class passengers. Fare distributions also varied between male and female passengers.

```python
# 6. Scatter Plot: Age vs Fare

plt.figure(figsize=(8,5))

sns.scatterplot(data=titanic, x='age', y='fare', hue='survived', alpha=0.7)

plt.title("Age vc Fare by Survival Status")
plt.xlabel("Age")
plt.ylabel("Fare")
plt.legend(title="Survived (0 = No, 1 = Yes)")
plt.tight_layout()

plt.show()
```

Observation:

Passengers who paid higher fares appear to have had better survival outcomes overall. However, survival was influenced by several factors, not just fare or age.
